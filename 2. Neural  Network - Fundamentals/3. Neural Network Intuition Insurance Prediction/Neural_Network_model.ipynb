```python
import torch
import torch.nn as nn
```

```python
model = nn.Sequential(
    nn.Linear(5, 8),   # 5 input features → 8 neurons
    nn.ReLU(),

    nn.Linear(8, 4),   # 8 → 4 neurons
    nn.ReLU(),

    nn.Linear(4, 1),   # 4 → 1 output
    nn.Sigmoid()
)

print(model)
```

```python
x = torch.tensor([
    [40.0, 3.0, 800000.0, 10.0, 1200000.0]
])
```

```python
prediction = model(x)
```

```python
print(prediction)
```
