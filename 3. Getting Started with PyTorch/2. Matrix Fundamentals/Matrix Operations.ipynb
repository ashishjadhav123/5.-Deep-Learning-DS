```python
import numpy as np
```

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])
```

## Addition

```python

C = A + B

print(C)
```

## Subtraction

```python
D = B - A

print(D)
```

## Scalar Multiplication

```python
print(2 * A)
```

## Element-wise Multiplication

```python
E = A * B

print(E)
```

## Matrix Multiplication

```python
F = A @ B

print(F)
```

```python
np.dot(A, B)
```

```python
np.matmul(A, B)
```

```python
A.T
```

```python
np.transpose(A)
```
