```python
import torch
```

#### Gradient for a single input

```python
material_cost = torch.tensor(10, requires_grad=True, dtype=torch.float64)
labor_cost = torch.tensor(15, requires_grad=True, dtype=torch.float64)
```

```python
print(material_cost, labor_cost)
```

```python
total_cost = 3 * material_cost**2 + 5*labor_cost + 100
total_cost
```

```python
total_cost.backward()
```

```python
material_cost.grad, labor_cost.grad
```

#### Disable gradient computation

```python
x = torch.tensor(4, requires_grad=True, dtype=torch.float16)
y = x**2 + 5

print("Outside no_grad: ", y.requires_grad)
```

```python
y.backward()
print(x.grad)
```

```python
with torch.no_grad():
    y = x**2 + 5
    print("Inside no_grad: ", y.requires_grad)
```

#### Gradient for a vector

```python
x = torch.tensor([1,2,3],requires_grad=True,dtype=torch.float16)

y = 2 * x ** 3 + 7 
y
```

```python
result = y.sum()
result
```

```python
result.backward()
```

```python
x.grad
```
