```python
import torch
import torch.nn as nn
```

```python
criterion = nn.MSELoss()

y_pred = torch.tensor([0.9])
y_true = torch.tensor([0], dtype=torch.float)

loss = criterion(y_pred, y_true)
print(f"MSE Loss: {loss.item():0.2f}")
```

```python
criterion = nn.BCELoss()

y_pred = torch.tensor([0.9])
y_true = torch.tensor([0], dtype=torch.float)

loss = criterion(y_pred, y_true)
print(f"MSE Loss: {loss.item():0.2f}")
```
