```python
import torch
import torch.nn as nn
```

```python
loos_func = nn.CrossEntropyLoss()

y_pred = torch.tensor([[0.1, 0.2, 0.7]]) # Dog (0), Cat (1), Koala (2)
y_true = torch.tensor([2])

loss = loos_func(y_pred, y_true)

print(f"Cross Entorpy loss: {loss.item():0.2f}")
```
