```python
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim

from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv(r"bonus.csv")
df.head()
```

```python
X = df.drop(columns=['bonus', 'employee_id']).values
y = df[['bonus']].values

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42, test_size=0.2)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
X_train_tensor = torch.tensor(X_train, dtype=torch.float32)
y_train_tensor = torch.tensor(y_train, dtype=torch.float32)
X_test_tensor = torch.tensor(X_test, dtype=torch.float32)
y_test_tensor = torch.tensor(y_test, dtype=torch.float32)
```

```python
class BonusPredictor(nn.Module):
    def __init__(self, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.network = nn.Sequential(
            nn.Linear(3, 1)     # 3 Input feature, 1 Output neuron
        )

    def forward(self, x):
        return self.network(x)
```

```python
model = BonusPredictor()

criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.005)
```

```python
X_train_tensor[:2]
```

```python
epochs = 5000

for epoch in range(epochs):
    # STEP:1  Forward Pass
    prediction = model(X_train_tensor)

    loss = criterion(prediction, y_train_tensor)

    # STEP:2  Backward Pass
    optimizer.zero_grad()

    loss.backward()

    # STEP:  Update weights
    optimizer.step()

    if (epoch + 1) % 100 == 0:
        print(f"Epoch: [{epoch + 1}/{epochs}], Loss: {loss.item():.2f}")
```

```python
model.eval()

with torch.no_grad():
    test_pred = model(X_test_tensor)
    test_loss = criterion(test_pred, y_test_tensor)

print(f"{test_loss.item():0.2f}")    
```
