```python
import torch
from torchvision import datasets
from torch.utils.data import DataLoader
from torchvision.transforms import ToTensor
```

```python
training_data = datasets.FakeData(
    size=2000,             # Number of synthetic samples
    image_size=(3, 32, 32),# Image dimensions (Channels, Height, Width)
    num_classes=10,        # Number of output classes
    transform=ToTensor()
)
```

```python
len(training_data)
```

```python
# Test Data (notice train=False)
test_data = datasets.FakeData(
    size=500,              # Smaller size for evaluation
    image_size=(3, 32, 32),
    num_classes=10,
    transform=ToTensor()
)
```

```python
len(test_data)
```

```python
train_loader = DataLoader(
    training_data,
    batch_size=64,
    shuffle=True
)
```

```python
images, labels = next(iter(train_loader))
print(f"Batch image tensor shape: {images.shape}")  # Expected: torch.Size([64, 3, 32, 32])
print(f"Batch label tensor shape: {labels.shape}")
```
